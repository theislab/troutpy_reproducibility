```python
# Segmentation methods to compare: default segmentation lives directly in `rois`,
# `rois_proseg` and `rois_segger` are the other two segmentation methods
# (see 12_segmentation_method_consistency.ipynb)
base_dir = "/lustre/groups/ml01/datasets/projects/2025_sergio_troutpy/rois/"
segmentations = ["rois", "rois_proseg", "rois_segger"]
seg_suffixes = {"rois": "", "rois_proseg": "_proseg", "rois_segger": "_segger"}

sdata_by_seg = {}
results_by_seg = {}

for seg in segmentations:
    # Technology 1: Xenium v1
    sdata1 = sd.read_zarr(f"{base_dir}{seg}/xenium_opsc_v1/data_with_source_score.zarr")
    results1 = sdata1["xrna_metadata"].var

    # Technology 2: Xenium 5k
    sdata2 = sd.read_zarr(f"{base_dir}{seg}/xenium_opsc_5k/data_with_source_score.zarr")
    results2 = sdata2["xrna_metadata"].var

    sdata_by_seg[seg] = (sdata1, sdata2)
    results_by_seg[seg] = (results1, results2)
```

```python
for seg in segmentations:
    sdata1, sdata2 = sdata_by_seg[seg]
    tl.quantify_overexpression(sdata1, codeword_key="control_probe", control_codewords=[True])
    tl.quantify_overexpression(sdata2, codeword_key="control_probe", control_codewords=[True])
    # quantify_overexpression reassigns sdata["xrna_metadata"].var to a new joined
    # DataFrame rather than mutating it in place, so results_by_seg must be refreshed
    # here rather than reusing the references captured before quantification
    results_by_seg[seg] = (sdata1["xrna_metadata"].var, sdata2["xrna_metadata"].var)
```

```python
merged_info_by_seg = {}
for seg in segmentations:
    results1, results2 = results_by_seg[seg]
    results1_filt = results1[results1.index.isin(results2.index)]
    results2_filt = results2[results2.index.isin(results1.index)]
    results2_filt = results2_filt.loc[results1_filt.index]
    merged_info_by_seg[seg] = results1_filt.merge(
        results2_filt, left_index=True, right_index=True, suffixes=("_v1", "_5k"), how="inner"
    )
```

```python
for seg in segmentations:
    merged_info_by_seg[seg].fillna(0, inplace=True)
```

```python
merged_info_by_seg["rois"].columns
```

```python
# Exclude genes with logfoldchange_over_noise <= 0 in either (or both) Xenium v1 and Xenium 5k
for seg in segmentations:
    mi = merged_info_by_seg[seg]
    mi = mi[mi["logfoldchange_over_noise_v1"] > 0]
    mi = mi[mi["logfoldchange_over_noise_5k"] > 0]
    merged_info_by_seg[seg] = mi
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
import matplotlib
import os

# Set font type for PDF and PostScript files
matplotlib.rcParams['pdf.fonttype'] = 42
matplotlib.rcParams['ps.fonttype'] = 42

# Define the saving path for the figures
saving_path = "/lustre/groups/ml01/datasets/projects/2025_sergio_troutpy/rois/figures/"

# Create the directory if it doesn't exist
if not os.path.exists(saving_path):
    os.makedirs(saving_path)
    print(f"Created directory: {saving_path}")

metrics = ["extracellular_proportion", "in_out_spearmanR", "moran_I", "logfoldchange_over_noise"]
num_genes_to_annotate = 2 # Number of top/bottom genes to annotate per axis

for seg in segmentations:
    merged_info = merged_info_by_seg[seg]
    seg_suffix = seg_suffixes[seg]
    seg_label = seg.replace("_", " ").title()

    for metric in metrics:
        fig, ax = plt.subplots(figsize=(7, 7)) # Slightly increased figure size for more annotations

        # Scatter Plot - Removed alpha for no transparency
        ax.scatter(
            merged_info[metric + "_v1"],
            merged_info[metric + "_5k"],
            s=15,
            color="#336B87",  # Darker blue for scatter points
            edgecolors='w',
            linewidths=0.5,
            zorder=2 # Ensure scatter points are above the grid (if any)
        )

        # Labels and Title (Updated to reflect Xenium v1 vs Xenium 5k)
        ax.set_xlabel(f"{metric.replace('_', ' ').title()} (Xenium v1)", fontsize=12)
        ax.set_ylabel(f"{metric.replace('_', ' ').title()} (Xenium 5k)", fontsize=12)

        # Calculate Cosine Similarity
        vec1 = merged_info.loc[:, metric + "_v1"].values.reshape(1, -1)
        vec2 = merged_info.loc[:, metric + "_5k"].values.reshape(1, -1)
        similarity = cosine_similarity(vec1, vec2)[0][0]

        # Title with Cosine Similarity and segmentation method
        ax.set_title(f"Comparison of {metric.replace('_', ' ').title()} ({seg_label})\nCosine Similarity: {similarity:.2f}", fontsize=14)

        # Removed Grid and Simplified Axes
        ax.grid(False) # Explicitly turn off the grid

        # Remove top and right spines to leave only x and y axes
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)

        ax.tick_params(axis='x', labelsize=10) # Adjusted tick label font size
        ax.tick_params(axis='y', labelsize=10) # Adjusted tick label font size

        # Add a diagonal line for perfect agreement
        min_val = min(merged_info[metric + "_v1"].min(), merged_info[metric + "_5k"].min())
        max_val = max(merged_info[metric + "_v1"].max(), merged_info[metric + "_5k"].max())
        ax.plot([min_val, max_val], [min_val, max_val], color='#A8201A', linestyle='--', alpha=0.7, label='Perfect Agreement', zorder=1)

        # --- Annotate Top and Bottom Genes ---
        annotated_genes = set() # Keep track of genes already annotated to avoid duplicates

        # Annotate Top genes for Xenium v1 axis
        top_xenium_v1_genes = merged_info.sort_values(by=metric + "_v1", ascending=False).head(num_genes_to_annotate)
        for gene_name, row in top_xenium_v1_genes.iterrows(): # Iterate using gene_name from index
            if gene_name not in annotated_genes:
                ax.annotate(
                    gene_name, # Use gene_name from index
                    (row[metric + "_v1"], row[metric + "_5k"]),
                    textcoords="offset points",
                    xytext=(5, 5), # Offset text slightly from the point
                    ha='left', # Horizontal alignment
                    va='bottom', # Vertical alignment
                    fontsize=10,
                    color='black',
                    arrowprops=dict(arrowstyle='-', connectionstyle='arc3,rad=0.2', color='gray', lw=1)
                )
                annotated_genes.add(gene_name)

        # Annotate Bottom genes for Xenium v1 axis
        bottom_xenium_v1_genes = merged_info.sort_values(by=metric + "_v1", ascending=True).head(num_genes_to_annotate)
        for gene_name, row in bottom_xenium_v1_genes.iterrows(): # Iterate using gene_name from index
            if gene_name not in annotated_genes:
                ax.annotate(
                    gene_name, # Use gene_name from index
                    (row[metric + "_v1"], row[metric + "_5k"]),
                    textcoords="offset points",
                    xytext=(5, -10), # Offset text slightly below the point
                    ha='left',
                    va='top', # Vertical alignment
                    fontsize=10,
                    color='black',
                    arrowprops=dict(arrowstyle='-', connectionstyle='arc3,rad=-0.2', color='gray', lw=1)
                )
                annotated_genes.add(gene_name)

        # Annotate Top genes for Xenium 5k axis
        top_xenium_5k_genes = merged_info.sort_values(by=metric + "_5k", ascending=False).head(num_genes_to_annotate)
        for gene_name, row in top_xenium_5k_genes.iterrows(): # Iterate using gene_name from index
            if gene_name not in annotated_genes:
                ax.annotate(
                    gene_name, # Use gene_name from index
                    (row[metric + "_v1"], row[metric + "_5k"]),
                    textcoords="offset points",
                    xytext=(-5, 5), # Offset text slightly from the point
                    ha='right', # Horizontal alignment
                    va='bottom',
                    fontsize=10,
                    color='black',
                    arrowprops=dict(arrowstyle='-', connectionstyle='arc3,rad=-0.2', color='gray', lw=1)
                )
                annotated_genes.add(gene_name)

        # Annotate Bottom genes for Xenium 5k axis
        bottom_xenium_5k_genes = merged_info.sort_values(by=metric + "_5k", ascending=True).head(num_genes_to_annotate)
        for gene_name, row in bottom_xenium_5k_genes.iterrows(): # Iterate using gene_name from index
            if gene_name not in annotated_genes:
                ax.annotate(
                    gene_name, # Use gene_name from index
                    (row[metric + "_v1"], row[metric + "_5k"]),
                    textcoords="offset points",
                    xytext=(-5, -10), # Offset text slightly below the point
                    ha='right',
                    va='top', # Vertical alignment
                    fontsize=10,
                    color='black',
                    arrowprops=dict(arrowstyle='-', connectionstyle='arc3,rad=0.2', color='gray', lw=1)
                )
                annotated_genes.add(gene_name)
        # --- End Annotate Top and Bottom Genes ---

        ax.legend() # Ensure legend is shown after all plot elements

        plt.tight_layout()
        # Save the plot as a PDF
        plt.savefig(f"{saving_path}/{metric}_xenium_v1_vs_5k_comparison{seg_suffix}.pdf")
        plt.show()
```
